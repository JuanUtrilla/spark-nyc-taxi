# NYC Taxi Dataset

Viajes en taxi en la ciudad de Nueva York (sólo taxis amarillos) en los años 2015 y 2016.

**Taxis amarillos**: estos son los famosos taxis amarillos de la ciudad de Nueva York que prestan servicio exclusivamente mediante señas en la calle. El número de taxis está limitado por la cantidad finita de licencias emitidas por la TLC. Se accede a este modo de transporte colocándose en la calle y llamando con la mano a un taxi disponible. Las recogidas no están prearregladas.

**Vehículos de alquiler (FHV)**: el servicio FHV se accede mediante una reserva previa con un despachador o una compañía de limusinas. Estos FHV no están autorizados a recoger pasajeros mediante señas en la calle, ya que esos viajes no se consideran pre-reservados.

**Taxis verdes**: Street Hail Livery (SHL). El programa SHL permitirá a los propietarios de vehículos de alquiler (VTC) licenciar y equipar sus vehículos con la identidad de taxi verde del distrito, instalar taxímetros y máquinas para tarjetas de crédito y, en última instancia, obtener el derecho a aceptar señas en la calle además de viajes pre-reservados.

En este dataset se consideran únicamente los datos de taxis amarillos de los meses de enero de 2015 y de enero a marzo de 2016.

```python
#!pip install folium==0.20.0 fsspec==2024.10.0 geopandas==1.1.2
# (automáticamente instalará shapely 2.1.2 y pyproj 3.7.2)

# dbutils.library.restartPython()
```

```python
from pyspark.sql import functions as F

# --- Databricks Free Edition: rellena con los datos de tu storage account ---
storage_account_name = "TU_STORAGE_ACCOUNT"
storage_account_key = "TU_API_KEY_DEL_STORAGE"
container_name = "TU_CONTAINER"

taxi_csv_path = f"abfss://{container_name}@{storage_account_name}.dfs.core.windows.net/yellow_tripdata_2015-01.csv"

nyc_trips_df = (
    spark.read
    .option("header", "true")
    .option("inferSchema", "true")
    .option("adls.account.name", storage_account_name)
    .option(f"fs.azure.account.key.{storage_account_name}.dfs.core.windows.net", storage_account_key)
    .csv(taxi_csv_path)
    .filter(
        (F.col("trip_distance") > 0)
        & (F.col("total_amount") > 0)
        & (F.col("fare_amount") > 0)
        & (F.col("total_amount") >= F.col("fare_amount"))
    )
    .withColumn(
        "trip_duration_s",
        F.timestamp_diff("SECOND", F.col("tpep_pickup_datetime"), F.col("tpep_dropoff_datetime")),
    )
    .filter(
        (F.col("trip_distance") >= 0.62)
        & (F.col("trip_duration_s") >= 120)
    )
    .withColumn("pickup_hour", F.hour("tpep_pickup_datetime"))
    .withColumn("dropoff_hour", F.hour("tpep_dropoff_datetime"))
    .withColumn("day_of_week", F.dayofweek("tpep_pickup_datetime"))
    .withColumn(
        "day_of_week",
        F.when(F.col("day_of_week") == 1, "domingo")
         .when(F.col("day_of_week") == 2, "lunes")
         .when(F.col("day_of_week") == 3, "martes")
         .when(F.col("day_of_week") == 4, "miercoles")
         .when(F.col("day_of_week") == 5, "jueves")
         .when(F.col("day_of_week") == 6, "viernes")
         .when(F.col("day_of_week") == 7, "sabado"),
    )
    .withColumn("speed", F.round(F.col("trip_distance") / (F.col("trip_duration_s") / 3600), 2))
    .withColumn("tip_perc", F.col("tip_amount") / F.col("total_amount"))
    .cache()
)
```

## Ejercicio 2 (2 puntos)

Se pide crear una función `compute_boxplot_stats(df, col_name)` que reciba como argumentos un DF de Spark (en el que asumimos que existen las columnas `pickup_hour` y `day_of_week` como string) y un string con el nombre de una columna numérica `col_name`, y devuelva un nuevo DF de Spark con una fila por cada hora del día (0-23) y 5 columnas por cada día de la semana (en total 36 columnas, incluyendo la columna con la hora del día), donde cada una de las 5 columnas son los valores con los que se construye un boxplot. Dichas columnas serán, para cada grupo, los números reales q1, median, q3, whisker_low, whisker_high que resultan de aplicar funciones de agregación.

Antes de agrupar:

* La función debe quedarse solo con las filas en las que `col_name` no sea NULL.
* A continuación, se recomienda calcular q1, median y q3 con las funciones `F.percentile_approx(col_name, 0.25)` o 0.75, y F.median, **aplicadas sobre una ventana** por hora y día de la semana.
* Una vez calculadas, con el fin de calcular los bigotes:
  * El bigote inferior real se define como el mínimo valor existente en la columna que sea mayor o igual a q1 - 1.5*(q3 - q1). Cualquier valor menor que ese bigote inferior se consideraría un outlier y no los vamos a representar.
  * El bigote superior real se define como el máximo valor existente en la columna que sea menor o igual a q3 + 1.5*(q3 - q1). Cualquier valor mayor que ese bigote superior se consideraría un outlier y no los vamos a representar.
  * Primero, crear dos columnas booleanas `valido_bigote_inferior`, `valido_bigote_superior` que comprueben, respectivamente, si cada dato es mayor o igual que el límite inferior, y si cada dato es menor o igual al límite superior.
  * Después, agrupando de la manera adecuada, dentro de la función agg, usar estas columnas para calcular el valor real de los bigotes con `F.min` y `F.max` respectivamente, **agregando solo las filas de los valores válidos** (es decir, aquellas donde la columna de validación calculada en el paso anterior es True). Puedes usar `F.min(F.when(..., F.col(col_name))).alias('whisker_low')` sin usar otherwise, ya que F.when devolverá NULL cuando no se cumple la condición pero las funciones `F.min` y `F.max` automáticamente ignoran esos nulos al calcular. Análogo con `F.max` para `whisker_high`. Para q1, median y q3, ya estarán calculado los valores en la ventana que se aplicó previamente, con lo que basta utilizar `F.first` para obtenerlos en la agregación final.

* La función debe tener el código lo más conciso posible. Si está correctamente implementada, el número de líneas de código efectivo está entre 15 y 20 (son 6 transformaciones más el groupBy).
* Invocar a la función creada con el DF de taxis y la columna `trip_distance`, guardando el resultado en la variable `stats_df` que debe quedar cacheada.
* Si está correctamente creado, se puede invocar a la función `plot_boxplot_facets` de aux_func para dibujar los boxplots facetados por hora del día y día de la semana.

```python

from pyspark.sql import Window
from pyspark.sql import DataFrame as DF

def compute_boxplot_stats(df: DF, col_name: str) -> DF:
    w = Window.partitionBy("pickup_hour", "day_of_week")
    df_stats = (
        df.filter(F.col(col_name).isNotNull())
        .withColumn("q1", F.percentile_approx(col_name, 0.25).over(w))
        .withColumn("q3", F.percentile_approx(col_name, 0.75).over(w))
        .withColumn("median", F.median(col_name).over(w))
        .withColumn("valido_bigote_inferior", F.col(col_name) >= F.col("q1") - 1.5 * (F.col("q3") - F.col("q1")))
        .withColumn("valido_bigote_superior", F.col(col_name) <= F.col("q3") + 1.5 * (F.col("q3") - F.col("q1")))
    )
    return (
        df_stats.groupBy("pickup_hour")
        .pivot("day_of_week")
        .agg(
            F.first("q1").alias("q1"),
            F.first("median").alias("median"),
            F.first("q3").alias("q3"),
            F.min(F.when(F.col("valido_bigote_inferior"), F.col(col_name))).alias("whisker_low"),
            F.max(F.when(F.col("valido_bigote_superior"), F.col(col_name))).alias("whisker_high"),
        )
    )

stats_df = compute_boxplot_stats(nyc_trips_df, "trip_distance").cache()
```

```python
assert stats_df.count() == 24 and len(stats_df.columns) == 36 and "pickup_hour" in stats_df.columns and "lunes_median" in stats_df.columns
stats_pd = compute_boxplot_stats(nyc_trips_df, 'trip_distance').sort("pickup_hour").toPandas()
assert round(stats_pd.lunes_q1[0], 1) == 1.5
assert round(stats_pd.martes_q3[10], 2) == 2.7

from aux_func import plot_boxplot_facets, limpia_lineas_codigo

plot_boxplot_facets(stats_df, "trip_distance")

import inspect
lineas, _ = inspect.getsourcelines(compute_boxplot_stats)
lineas_limpias = limpia_lineas_codigo(lineas)  # número de líneas de código efectivo
assert(len(lineas_limpias) < 30)
```

```python
from typing import List
import aux_func
from importlib import reload
reload(aux_func)

from aux_func import plot_week_hour_heatmap
from pyspark.sql import DataFrame as DF

import seaborn as sns

def medianas_y_conteo(df: DF, median_cols: List[str]) -> DF:
    agg_exprs = [F.count(F.lit(1)).alias("n_trips")] + [F.median(c).alias(c) for c in median_cols]
    return (
        df.groupBy("day_of_week")
        .pivot("pickup_hour")
        .agg(*agg_exprs)
    )

df_with_amount_per_mile = nyc_trips_df.withColumn(
    "total_amount_per_mile",
    F.when(F.col("trip_distance") != 0, F.col("total_amount") / F.col("trip_distance")),
)

counts_and_distances_df = medianas_y_conteo(
    df_with_amount_per_mile, ["trip_distance", "total_amount_per_mile", "speed"]
).cache()

cols_trip_distance = [F.col("day_of_week")] + [F.col(f"{h}_trip_distance").alias(str(h)) for h in range(24)]
cols_total_amount_per_mile = [F.col("day_of_week")] + [F.col(f"{h}_total_amount_per_mile").alias(str(h)) for h in range(24)]
cols_n_trips = [F.col("day_of_week")] + [F.col(f"{h}_n_trips").alias(str(h)) for h in range(24)]
cols_speed = [F.col("day_of_week")] + [F.col(f"{h}_speed").alias(str(h)) for h in range(24)]

trip_distance_by_hour_pd = counts_and_distances_df.select(*cols_trip_distance).toPandas()
total_amount_by_hour_pd = counts_and_distances_df.select(*cols_total_amount_per_mile).toPandas()
n_trips_by_hour_pd = counts_and_distances_df.select(*cols_n_trips).toPandas()
speed_by_hour_pd = counts_and_distances_df.select(*cols_speed).toPandas()
```

Vamos a encontrar a qué distrito pertenece cada punto de recogida y de destino de los viajes en taxi. Para calcular el distrito de recogida y el distrito de destino de cada viaje, debes seguir estos pasos:

1. Añadir a `nyc_gdf` (GeoDataFrame, que se comporta como Pandas) una columna `district_geom` con la geometría de cada distrito en formato WKT (Well-Known Text), usando el método `apply(lambda g: g.wkt)` de Pandas sobre la columna `geometry`.
2. Crear un DF de Spark a partir del GeoDataFrame, tal como hemos visto en clase que se hace cuando partimos de un DF de Pandas, pero seleccionando únicamente las columnas `district` y `district_geom`. Encadena la creación con una transformación para reemplazar la columna `district_geom` por el resultado de aplicar la función `ST_GeomFromWKT` de Sedona para que la columna sean objetos de geometría. Como esta función solamente está disponible en la parte de SQL pura de Spark, necesitarás invocarla con `F.expr("ST_GeomFromWKT(...)")`. El resultado se guardará en la variable `districts_df`.
3. Añadir las columnas `pickup_point` y `dropoff_point` al DF de `nyc_trips_df`, con la función `ST_Point(longitud, latitud)` de Sedona aplicada a las coordenadas de longitud y latitud de recogida y destino, respectivamente. De nuevo tendrás que usar `F.expr`. Ten cuidado con el orden de latitud y longitud, ya que la función `ST_Point` espera primero la longitud (X) y luego la latitud (Y). 4. Crear dos DF y guardarlos en las variables `pickup_districts` y `dropoff_districts`, creadas a partir de `districts_df`, simplemente renombrando la columna `district` a `pickup_district` en el primer caso, y a `dropoff_district` en el segundo caso, para que luego se puedan distinguir en los joins.
5. Realiza un join espacial entre `trips_with_points` y `pickup_districts` para encontrar el distrito de recogida de cada viaje, usando la función `ST_Contains(geometría, punto)` como condición de join (con `F.expr(ST_Contains(...))`) *asegurando que nunca se pierdan viajes*. Recuerda indicar como tabla de la derecha `F.broadcast(pickup_districts)` por ser una tabla muy pequeña que queremos que se distribuya a los executors. El resultado debe encadenarse con la eliminación de las columnas `pickup_point` y `district_geom`, y después con otro join espacial con `dropoff_districts` para encontrar el distrito de destino de cada viaje, usando la misma función `ST_Contains`, y eliminando de nuevo `dropoff_point` y `district_geom`. El resultado final debe quedar almacenado en la variable `nyc_trips_with_districts` que tendrá dos columnas nuevas: `pickup_district` y `dropoff_district`, con el nombre del distrito comunal correspondiente a cada punto de recogida y destino.
  * Ya sabemos que saldrán algunos valores nulos en estas columnas nuevas, pero no debes eliminarlos por el momento, ya que no influyen en el conteo por barrios que vamos a hacer después.
6. Ahora que tenemos el distrito de origen y destino de cada viaje, crea un nuevo DF de Spark con tantas filas como distritos de recogida, y tantas columnas como días de la semana más una (el distrito), con el número total de viajes que se **iniciaron** en cada distrito cada día de la semana, y pasa el resultado a un DF de Pandas que debe quedar guardado en `trips_per_district_pd`.

```python

# -----------------------------------------------------------
# Descomentar SOLAMENTE si estás utilizando Spark en tu PC local
# -----------------------------------------------------------
# from sedona.spark.register import SedonaRegistrator
# sedona.spark.register.SedonaRegistrator.registerAll(spark)
# -----------------------------------------------------------

nyc_gdf["district_geom"] = nyc_gdf["geometry"].apply(lambda g: g.wkt)

districts_df = (
    spark.createDataFrame(nyc_gdf[["district", "district_geom"]])
    .withColumn("district_geom", F.expr("ST_GeomFromWKT(district_geom)"))
)

trips_with_points = (
    nyc_trips_df
    .withColumn("pickup_point", F.expr("ST_Point(pickup_longitude, pickup_latitude)"))
    .withColumn("dropoff_point", F.expr("ST_Point(dropoff_longitude, dropoff_latitude)"))
)

pickup_districts = districts_df.withColumnRenamed("district", "pickup_district")
dropoff_districts = districts_df.withColumnRenamed("district", "dropoff_district")

nyc_trips_with_districts = (
    trips_with_points
    .join(
        F.broadcast(pickup_districts),
        on=F.expr("ST_Contains(district_geom, pickup_point)"),
        how="left",
    )
    .drop("pickup_point", "district_geom")
    .join(
        F.broadcast(dropoff_districts),
        on=F.expr("ST_Contains(district_geom, dropoff_point)"),
        how="left",
    )
    .drop("dropoff_point", "district_geom")
)

trips_pickup = (
    nyc_trips_with_districts
    .groupBy("pickup_district")
    .pivot("day_of_week")
    .agg(F.count(F.lit(1)))
)

trips_per_district_pd = trips_pickup.toPandas()
```
